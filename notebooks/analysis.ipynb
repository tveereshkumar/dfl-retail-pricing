```python
import pandas as pd
df = pd.read_csv("../data/synthetic_data.csv")
df.head()
```

```python
import matplotlib.pyplot as plt

df_sku = df[df["sku"] == "SKU_1"]

plt.scatter(df_sku["price"], df_sku["demand"])
plt.xlabel("Price")
plt.ylabel("Demand")
plt.title("Price vs Demand")
plt.show()
```
